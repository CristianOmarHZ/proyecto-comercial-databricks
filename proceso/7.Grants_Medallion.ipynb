# 7.Grants_Medallion — Olist PROD

Otorga acceso a `grp_olist_ingenieria` a las capas Bronze, Silver y Golden; y solo lectura de Golden a `grp_olist_analistas`. Estos son permisos del catálogo, no sustituye los permisos del clúster o SQL Warehouse.


```python
# 7.Grants_Medallion — PRODUCCIÓN
# No crea usuarios ni grupos. Los grupos deben existir en la cuenta Databricks.
dbutils.widgets.text("catalogo", "catalog_prod")
catalogo = dbutils.widgets.get("catalogo").strip()
if catalogo != "catalog_prod":
    raise ValueError("Este notebook solo aplica grants en catalog_prod")

comandos = [
    "GRANT USE CATALOG ON CATALOG catalog_prod TO `grp_olist_ingenieria`",
    "GRANT USE CATALOG ON CATALOG catalog_prod TO `grp_olist_analistas`",
]
for capa in ["bronze", "silver", "golden"]:
    comandos.extend([
        f"GRANT USE SCHEMA ON SCHEMA catalog_prod.{capa} TO `grp_olist_ingenieria`",
        f"GRANT SELECT, MODIFY, CREATE TABLE ON SCHEMA catalog_prod.{capa} TO `grp_olist_ingenieria`",
    ])
comandos.extend([
    "GRANT USE SCHEMA ON SCHEMA catalog_prod.golden TO `grp_olist_analistas`",
    "GRANT SELECT ON SCHEMA catalog_prod.golden TO `grp_olist_analistas`",
])

for comando in comandos:
    spark.sql(comando)
    print("OK:", comando)
print("Grants de catalog_prod completados. No se alteraron tablas ni datos.")
```
